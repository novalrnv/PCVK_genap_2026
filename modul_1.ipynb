```python
import numpy as np
import pandas as pd
import cv2 as cv
from google.colab.patches import cv2_imshow # for image display
from skimage import io
from skimage import transform
from PIL import Image
import matplotlib.pylab as plt

# Membuat list untuk menyimpan url dari beberapa image
urls = ["https://iiif.lib.ncsu.edu/iiif/0052574/full/800,/0/default.jpg",
        "https://iiif.lib.ncsu.edu/iiif/0016007/full/800,/0/default.jpg",
        "https://images.unsplash.com/photo-1514888286974-6c03e2ca1dba?w=800"
        ]
# baca dan tampilkan image
# loop pada tiap url image, beberapa image dapat disimpan pada list
for url in urls:
  image = io.imread(url)                           #read image
  image = cv.resize(image, (0,0), fx=0.5, fy=0.5)  #resize image to half size
  image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)   #convert color to RGB
  final_frame = cv.hconcat((image, image_2))       #concatenate image
  cv2_imshow(final_frame)                          #show image
  print('\n')


tinggi = image_2.shape[0]
lebar = image_2.shape[1]
print("resolusi image: tinggi x lebar = ",tinggi," x ",lebar)
cv2_imshow(image_2)


image_1 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (lebar):
  image_2[int((tinggi)/2),y] = [255,255,255]

final_frame = cv.hconcat((image_1, image_2))
cv2_imshow(final_frame)

print("")

image_1 = cv.cvtColor(image, cv.COLOR_BGR2RGB)
image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

#membuat garis horizontal ditengah image
for y in range (tinggi):
  image_2[y, int((lebar)/2)] = [255,255,255]

final_frame = cv.hconcat((image_1, image_2))
cv2_imshow(final_frame)

print("")

image_1 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# membuat garis diagonal dari pojok kiri atas ke pojok kanan bawah
cv.line(image_2, (0, 0), (lebar - 1, tinggi - 1), (255, 255, 255), 1)

# membuat garis diagonal dari pojok kanan atas ke pojok kiri bawah
cv.line(image_2, (lebar - 1, 0), (0, tinggi - 1), (255, 255, 255), 1)

# menggabungkan gambar asli dan gambar yang sudah diberi garis silang
final_frame = cv.hconcat((image_1, image_2))

cv2_imshow(final_frame)

print("")

image_1 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# menentukan posisi tengah gambar
tengah_y = int(tinggi / 2)
tengah_x = int(lebar / 2)

# menentukan panjang garis
panjang_garis = 180

# menentukan titik awal dan titik akhir garis
x1 = int(tengah_x - panjang_garis / 2)
x2 = int(tengah_x + panjang_garis / 2)

# membuat garis horizontal putih di tengah gambar
cv.line(image_2, (x1, tengah_y), (x2, tengah_y), (255, 255, 255), 1)

# menggabungkan gambar asli dan gambar yang sudah diberi garis
final_frame = cv.hconcat((image_1, image_2))

# menampilkan hasil
cv2_imshow(final_frame)

print("")

image_1 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

image_2 = cv.cvtColor(image, cv.COLOR_BGR2RGB)

# menentukan posisi kotak
x1 = 150
y1 = 80

# menentukan ukuran kotak
lebar_kotak = 100
tinggi_kotak = 100

# membuat kotak berwarna putih
cv.rectangle(
    image_2, (x1, y1),
    (x1 + lebar_kotak, y1 + tinggi_kotak),
    (255, 255, 255), -1
)

# menggabungkan gambar asli dan gambar yang sudah diberi kotak
final_frame = cv.hconcat((image_1, image_2))

# menampilkan hasil
cv2_imshow(final_frame)
```
