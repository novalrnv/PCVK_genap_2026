```python
from PIL import Image
import matplotlib.pyplot as plt

# Membuka gambar
ktm1 = Image.open("/content/drive/MyDrive/PCVK/Jobsheet1/KTM1.jpeg")
ktm2 = Image.open("/content/drive/MyDrive/PCVK/Jobsheet1/KTM2.jpeg")

# Menampilkan gambar
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.imshow(ktm1)
plt.title("KTM1.jpg - Sebelum Crop")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(ktm2)
plt.title("KTM2.jpg - Setelah Crop")
plt.axis("off")

plt.show()
```

```python
print("Dimensi KTM1.jpg:", ktm1.size)
print("Dimensi KTM2.jpg:", ktm2.size)

print("")

pixel_ktm1 = ktm1.width * ktm1.height
pixel_ktm2 = ktm2.width * ktm2.height

print("Total pixel KTM1.jpg:", pixel_ktm1)
print("Total pixel KTM2.jpg:", pixel_ktm2)
```
