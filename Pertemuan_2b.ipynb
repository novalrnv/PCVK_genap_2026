```python
import cv2
import matplotlib.pyplot as plt
from skimage import io
import random

# Mengambil gambar random
url = "https://picsum.photos/800/600?random=1"
img = io.imread(url)

# Salinan gambar
img_box = img.copy()

# Ukuran gambar
tinggi, lebar = img_box.shape[:2]

# Warna kotak RGB
warna = [
    (255, 0, 0),      # Merah
    (0, 255, 0),      # Hijau
    (0, 0, 255),      # Biru
    (255, 255, 0),    # Kuning
    (255, 0, 255)     # Ungu
]

# Membuat 5 kotak
for i in range(5):

    # Posisi random
    x1 = random.randint(0, lebar - 100)
    y1 = random.randint(0, tinggi - 100)

    # Ukuran random
    lebar_kotak = random.randint(50, 200)
    tinggi_kotak = random.randint(50, 150)

    # Batas agar kotak tidak keluar gambar
    x2 = min(x1 + lebar_kotak, lebar - 1)
    y2 = min(y1 + tinggi_kotak, tinggi - 1)

    # Membuat kotak
    cv2.rectangle(
        img_box,
        (x1, y1),
        (x2, y2),
        warna[i],
        3
    )

# Menampilkan gambar
plt.figure(figsize=(10, 6))
plt.imshow(img_box)
plt.title("5 Kotak Random dengan Warna Berbeda")
plt.axis("off")

print("Tugas no.5")
print("")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt

# Tugas 8 dan 9, membuat kotak dan menampilkan nama file

# Membaca foto aktivitas
img = cv2.imread("/content/activity.jpeg")

# Mengubah BGR ke RGB
imgRGB = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)

# Ukuran gambar
tinggi, lebar = imgRGB.shape[:2]

# Membuat rectangle biru di bagian kiri bawah
x1 = 50
y1 = tinggi - 400
x2 = 300
y2 = tinggi - 100

cv2.rectangle(
    imgRGB,
    (x1, y1),
    (x2, y2),
    (0, 255, 0),
    3
)

# Menambahkan nama file di atas kotak
cv2.putText(
    imgRGB,
    "activity.jpeg",
    (x1, y1 - 20),
    cv2.FONT_HERSHEY_SIMPLEX,
    0.8,
    (0, 255, 255),
    2
)

# Menampilkan gambar
plt.figure(figsize=(8, 12))
plt.imshow(imgRGB)
plt.title("Nomor 8 dan 9, membuat kotak dan menampilkan nama file")
plt.axis("off")

print("Tugas no.8 dan 9")
print("")
plt.show()
```

```python
import cv2
import matplotlib.pyplot as plt
from google.colab import drive

path_ktm = '/content/drive/MyDrive/PCVK/Jobsheet1/KTM2.jpeg'
img_bgr = cv2.imread(path_ktm)

# Konversi BGR ke RGB untuk visualisasi Matplotlib
img_rgb = cv2.cvtColor(img_bgr, cv2.COLOR_BGR2RGB)
img_rgb = cv2.resize(img_rgb, (900, 600))

plt.figure(figsize=(10, 6))
plt.imshow(img_rgb)
plt.title("KTM Polinema - Asli")
plt.axis('off')
```

```python
ktm_k5 = img_rgb.copy()

# 1. Logo Polinema pojok kiri atas (Circle - Biru)
cv2.circle(ktm_k5, center=(75, 60), radius=42, color=(0, 102, 204), thickness=-1)

# 2. Tanggal / Angka Valid Thru (Rectangle - Merah Crimson)
cv2.rectangle(ktm_k5, pt1=(520, 380), pt2=(650, 425), color=(220, 20, 60), thickness=-1)

# 3. Tulisan D-IV pada prodi (Rectangle - Hijau Muda)
cv2.rectangle(ktm_k5, pt1=(190, 532), pt2=(250, 560), color=(46, 204, 113), thickness=-1)

plt.figure(figsize=(10, 6))
plt.imshow(ktm_k5)
plt.title("Kelompok 5: Sensor Logo POLINEMA, Valid Thru, dan D-IV")
plt.axis('off')
plt.show()
```
