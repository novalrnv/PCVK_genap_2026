```python
import cv2
import matplotlib.pyplot as plt
from google.colab.patches import cv2_imshow

img = cv2.imread("/content/drive/MyDrive/PCVK/Jobsheet1/KTM2.jpeg")
plt.imshow(img) #format warna BGR

img = cv2.imread("/content/drive/MyDrive/PCVK/Jobsheet1/KTM2.jpeg")
img = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
imgRGBsmall=cv2.resize(img,(400,250))
plt.imshow(imgRGBsmall)
```

```python
img_gray = cv2.imread("/content/drive/MyDrive/PCVK/Jobsheet1/KTM2.jpeg", cv2.IMREAD_GRAYSCALE)

# Tanpa cmap
plt.subplot(1,2,1)
plt.imshow(img_gray)
plt.title("Tanpa cmap")

# Dengan cmap
plt.subplot(1,2,2)
plt.imshow(img_gray, cmap="magma")
plt.title("Dengan cmap")

plt.show()
```
